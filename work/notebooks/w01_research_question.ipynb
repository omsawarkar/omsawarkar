## 3. Quick look at the data (2-3 real numbers)

The starter dataset contains 30,000 content items. Of these, 16,262 are marked as declining and have search impressions during the 90-day window. Within that group, 82 have not been updated for at least 180 days.

These numbers support exploring which pages should be reviewed first: the dataset contains many declining items, and combining signals can help narrow the review list. However, decline and time since an update do not prove that a page needs refreshing or that refreshing it will improve traffic. The 180-day cutoff is a provisional starting rule, not a proven threshold.

```python
import pandas as pd

# Load the starter dataset from your GitHub repository.
url = (
    "https://raw.githubusercontent.com/omsawarkar/omsawarkar/"
    "main/data/raw/content_refresh_anonymized.csv"
)
df = pd.read_csv(url)

# Find pages marked as declining that have search impressions.
declining_visible = (
    (df["trend_direction"] == "down")
    & (df["impressions_90d"] > 0)
)

# Among those pages, find ones not updated for at least 180 days.
older_declining = (
    declining_visible
    & (df["days_since_last_update"] >= 180)
)

print(f"Total content items: {len(df):,}")
print(
    "Declining items with search impressions: "
    f"{declining_visible.sum():,}"
)
print(
    "Of these, not updated for at least 180 days: "
    f"{older_declining.sum():,}"
)
```
